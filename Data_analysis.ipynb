### Merge files


```python
import pandas as pd
import numpy as np
import xy
```

```python
files = [
    "Data_specialist",
    "Data_Analyst",
    "Data_Scientist",
    "machine_learning",
    "Data_Analys_fa",
    "Data_Scientist_fa"
]
```

```python
# Read all and store in dictionary
dataframes = {}
for file in files:
    dataframes[file] = pd.read_excel(f"Data/{file}.xlsx", index_col=0)

# Merge all into one DataFrame
data = pd.concat(dataframes, ignore_index=True)
```

```python
# Initial inspection
print("Dataset Shape:", data.shape)
print("\nColumns:", data.columns.tolist())
print("\nMissing Values:\n", data.isnull().sum())
```

```python
## Make province and city columns from location column

def parse_persian_location(loc_str):
    if pd.isna(loc_str):
        return pd.Series([None, None])
    
    # Split by Arabic comma (،) or standard comma (,)
    parts = [part.strip() for part in str(loc_str).replace('،', ',').split(',')]
    
    if len(parts) >= 2:
        return pd.Series([parts[0], parts[1]]) # [Province, City]
    elif len(parts) == 1:
        return pd.Series([parts[0], None])
    return pd.Series([None, None])

# Apply parsing
data[['province', 'city']] = data['location'].apply(parse_persian_location)

# Removing location column
data.drop(columns="location", inplace=True)
```

```python
import plotly.express as px

# 1. Aggregate counts by Province
province_counts = data['province'].value_counts().reset_index()
province_counts.columns = ['province', 'job_count']

# Filter out nulls/unspecified locations
province_counts = province_counts.dropna().head(10)

# 2. Plot horizontal bar chart
fig_province = px.bar(
    province_counts,
    x='job_count',
    y='province',
    orientation='h',
    title='<b>Top 10 Iranian Provinces by Related Job Postings</b>',
    labels={'job_count': 'Number of Job Postings', 'province': 'Province'},
    color='job_count',
    color_continuous_scale='Blues',
    text='job_count'
)

# Reverse y-axis so top province is at the top
fig_province.update_layout(
    yaxis={'categoryorder': 'total ascending'},
    showlegend=False,
    template='plotly_white'
)

fig_province.show()
```

```python
# Location Variable
## 1. Aggregate counts by Province and City
province_counts = (
    data.groupby('province')
    .size()
    .reset_index(name='job_count')
)

province = province_counts["province"]
job_count = np.array(province_counts["job_count"])

order = np.argsort(job_count)
province_sorted = province[order]
job_count_sorted = job_count[order]

chart = xy.column_chart(
    xy.column(province_sorted,
              job_count_sorted,
              color="#6e56cf",
              orientation='horizontal',
              corner_radius=4
              ),
    xy.x_axis(label="job_count"),
    xy.y_axis(label="province"),
    title="Iranian Provinces by Related Job Postings",
    width=1500,
    height=500,
)

chart
```
